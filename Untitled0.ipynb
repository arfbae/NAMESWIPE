```python
import gradio as gr
import pandas as pd
import hashlib, zipfile, os, shutil
from PIL import Image
from transformers import pipeline

# 1. Load the AI Captioner
print("Loading AI... please wait.")
captioner = pipeline("image-to-text", model="Salesforce/blip-image-captioning-base")

def process_images(input_zip):
    # --- Added logging for troubleshooting ---
    print(f"Received input_zip: {input_zip}")
    if input_zip:
        print(f"Input zip name: {input_zip.name}")
        print(f"Input zip type: {type(input_zip)}")
        if hasattr(input_zip, 'size'):
            print(f"Input zip size: {input_zip.size} bytes")
    else:
        print("Input zip is None or empty.")
    # --- End of logging ---

    if not input_zip: return None, None

    # Setup folders
    extract_path = "temp_images"
    output_path = "renamed_images"
    if os.path.exists(extract_path): shutil.rmtree(extract_path)
    if os.path.exists(output_path): shutil.rmtree(output_path)
    os.makedirs(output_path)

    # Unzip
    try:
        with zipfile.ZipFile(input_zip.name, 'r') as zip_ref:
            zip_ref.extractall(extract_path)
    except zipfile.BadZipFile:
        print(f"Error: The file {input_zip.name} is not a valid zip file.")
        return None, None # Return None for outputs if zip is bad
    except Exception as e:
        print(f"An unexpected error occurred while unzipping: {e}")
        return None, None

    data_log = []

    # Process files
    for filename in os.listdir(extract_path):
        if filename.lower().endswith(('.png', '.jpg', '.jpeg')):
            img_path = os.path.join(extract_path, filename)

            # AI Logic
            result = captioner(img_path)
            caption = result[0]['generated_text']

            # Simplified Labeling (ASAP version)
            label = caption.split()[0] # Takes the first word as category
            short_hash = hashlib.md5(filename.encode()).hexdigest()[:6]
            ext = os.path.splitext(filename)[1]
            new_name = f"{label}-{short_hash}{ext}"

            # Save renamed file
            shutil.copy(img_path, os.path.join(output_path, new_name))

            # Log data
            data_log.append({
                "Original": filename,
                "Renamed": new_name,
                "Label": label,
                "Caption": caption
            })

    # Create CSV and Zip
    df = pd.DataFrame(data_log)
    df.to_csv("audit_trail.csv", index=False)

    final_zip = "renamed_assets.zip"
    with zipfile.ZipFile(final_zip, 'w') as zipf:
        for file in os.listdir(output_path):
            zipf.write(os.path.join(output_path, file), file)

    return final_zip, "audit_trail.csv" # Return the paths to the created files
```

```python
# @title AI prompt cell

import ipywidgets as widgets
from IPython.display import display, HTML, Markdown,clear_output
from google.colab import ai

dropdown = widgets.Dropdown(
    options=[],
    layout={'width': 'auto'}
)

def update_model_list(new_options):
    dropdown.options = new_options
update_model_list(ai.list_models())

text_input = widgets.Textarea(
    placeholder='Ask me anything....',
    layout={'width': 'auto', 'height': '100px'},
)

button = widgets.Button(
    description='Submit Text',
    disabled=False,
    tooltip='Click to submit the text',
    icon='check'
)

output_area = widgets.Output(
     layout={'width': 'auto', 'max_height': '300px','overflow_y': 'scroll'}
)

def on_button_clicked(b):
    with output_area:
        output_area.clear_output(wait=False)
        accumulated_content = ""
        for new_chunk in ai.generate_text(prompt=text_input.value, model_name=dropdown.value, stream=True):
            if new_chunk is None:
                continue
            accumulated_content += new_chunk
            clear_output(wait=True)
            display(Markdown(accumulated_content))

button.on_click(on_button_clicked)
vbox = widgets.GridBox([dropdown, text_input, button, output_area])

display(HTML("""
<style>
.widget-dropdown select {
    font-size: 18px;
    font-family: "Arial", sans-serif;
}
.widget-textarea textarea {
    font-size: 18px;
    font-family: "Arial", sans-serif;
}
</style>
"""))
display(vbox)
```

```python
import gradio as gr

# Create the Gradio interface
interface = gr.Interface(
    fn=process_images, # Our image processing function
    inputs=gr.File(type="filepath", label="Upload Zip File containing Images"), # Input is a zip file
    outputs=[
        gr.File(label="Renamed Assets Zip (Download Me)"),
        gr.File(label="Audit Trail CSV (Download Me)")
    ],
    title="Image Renamer with AI Captioning",
    description="Upload a zip file containing images (PNG, JPG, JPEG). The AI will caption each image, rename it based on the caption, and provide a new zip file with renamed images and an audit trail CSV."
)

# Launch the interface
interface.launch(debug=True)
```
